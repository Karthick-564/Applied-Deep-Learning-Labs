```python
# Name: Karthick N
# Roll No: 24BAD055

import numpy as np
import matplotlib.pyplot as plt
```

```python
#Define Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)
```

```python
#Generate Input Range
x = np.linspace(-10, 10, 400)

y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)

dy_sigmoid = sigmoid_derivative(x)
dy_tanh = tanh_derivative(x)
dy_relu = relu_derivative(x)
```

```python
#Plot Sigmoid,tanh,Relu
plt.figure(figsize=(10, 6))
plt.plot(x, y_sigmoid, label='Sigmoid', color='blue')
plt.plot(x, y_tanh, label='Tanh', color='green')
plt.plot(x, y_relu, label='ReLU', color='red')
plt.title('Activation Functions')
plt.xlabel('Input (x)')
plt.ylabel('Output')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x, dy_sigmoid, label='Sigmoid Derivative', color='blue')
plt.plot(x, dy_tanh, label='Tanh Derivative', color='green')
plt.plot(x, dy_relu, label='ReLU Derivative', color='red')
plt.title('Gradient (Derivative) Behavior of Activation Functions')
plt.xlabel('Input (x)')
plt.ylabel('Derivative')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Comparison table
print("Comparison of Activation Functions")
print()
print("Sigmoid  -> Range: (0,1)   | Zero-Centered: No  | Vanishing Gradient: Yes")
print("Tanh     -> Range: (-1,1)  | Zero-Centered: Yes | Vanishing Gradient: Yes (less)")
print("ReLU     -> Range: [0,inf) | Zero-Centered: No  | Vanishing Gradient: No (x>0)")
print()
print("Computation Cost -> Sigmoid: High, Tanh: High, ReLU: Low")
print("Typical Use -> Sigmoid: Output layer (binary), Tanh/ReLU: Hidden layers")
```
